```python
import tkinter as tk

def c_to_f():
    c = c_entry.get()
    if c:
        c_result.config(text=f"{(float(c)*9/5)+32:.2f} °F")

def f_to_c():
    f = f_entry.get()
    if f:
        f_result.config(text=f"{(float(f)-32)*5/9:.2f} °C")

root = tk.Tk()
root.title("Temp Converter")
root.geometry("300x200")

# Celsius → Fahrenheit
tk.Label(root, text="Celsius to Fahrenheit").pack()
celsius_entry = tk.Entry(root); celsius_entry.pack()
tk.Button(root, text="Convert", command=c_to_f).pack()
celsius_result = tk.Label(root, text="Enter °C above"); celsius_result.pack()

# Fahrenheit → Celsius
tk.Label(root, text="Fahrenheit to Celsius").pack(pady=5)
farenheit_entry = tk.Entry(root); farenheit_entry.pack()
tk.Button(root, text="Convert", command=f_to_c).pack()
farenheit_result = tk.Label(root, text="Enter °F above"); farenheit_result.pack()

root.mainloop()
```
